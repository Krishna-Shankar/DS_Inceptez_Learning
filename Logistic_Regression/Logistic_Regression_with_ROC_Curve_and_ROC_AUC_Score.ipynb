# **Logistic Regression with ROC Curve and ROC AUC Score**

```python
# roc curve and auc
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import roc_curve
from sklearn.metrics import roc_auc_score
from matplotlib import pyplot
```

```python
#generate 2 class dataset
X, y = make_classification(n_samples = 1000, n_classes=2, random_state=1)
```

```python
#split the dataset into train and test
train_X, test_X, train_y, test_y = train_test_split(X, y, test_size=0.25, random_state=0)
```

```python
#created a dummy model with default 0 as output
dummy_model_prob = [0 for _ in range(len(test_y))]
dummy_model_prob
```

```python
#Lets create basic Logistic Regression Model
model = LogisticRegression()
model.fit(train_X, train_y)
```

```python
#Prediction based on probability
model_prob = model.predict_proba(test_X)
```

```python
model_prob
```

```python
#Lets focus only one postive value
model_prob = model_prob[:,1]
```

```python
#Lets calculate the scores
dummy_model_auc = roc_auc_score(test_y, dummy_model_prob)
model_auc = roc_auc_score(test_y, model_prob)
print(dummy_model_auc)
print(model_auc)
```

```python
#calculate ROC curves
dummy_fpr, dummy_tpr, _ = roc_curve(test_y, dummy_model_prob)
model_fpr, model_tpr, thresholds = roc_curve(test_y, model_prob)
```

```python
#plot the roc curve for the model
pyplot.plot(dummy_fpr, dummy_tpr, linestyle='--', label='Dummy Model')
pyplot.plot(model_fpr, model_tpr, marker='.', label='Logistic Regression')
pyplot.xlabel('False Positive Rate')
pyplot.ylabel('True Positive Rate')
pyplot.legend()
pyplot.show()
```

```python
#plot the roc curve for the model
import numpy as np
fig = pyplot.figure(figsize=(20,50))
pyplot.plot(dummy_fpr, dummy_tpr, linestyle='--', label='Dummy Model')
pyplot.plot(model_fpr, model_tpr, marker='.', label='Logistic Regression')
ax = fig.add_subplot(111)
for xyz in zip(model_fpr, model_tpr, thresholds):
    ax.annotate('%s' % np.round(xyz[2],2), xy=(xyz[0], xyz[1]))
pyplot.xlabel('False Positive Rate')
pyplot.ylabel('True Positive Rate')
pyplot.legend()
pyplot.show()
```
