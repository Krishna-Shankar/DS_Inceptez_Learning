```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
from sklearn.datasets import make_classification
```

```python
#create the dataset
X,y = make_classification(n_samples=1000, n_features=10, n_classes=2, random_state=30)
```

```python
X
```

```python
y
```

```python
pd.DataFrame(X)
```

```python
pd.DataFrame(y)
```

```python


from sklearn.model_selection import train_test_split
train_X, test_X, train_y, test_y = train_test_split(X,y,test_size=0.30,random_state=42)
```

```python

# Model training
from sklearn.linear_model import LogisticRegression
logistic_model = LogisticRegression()
```

```python
logistic_model.fit(train_X, train_y)
```

```python
y_pred_train = logistic_model.predict(train_X)
y_pred_test = logistic_model.predict(test_X)
print(y_pred_train)
print(y_pred_test)
```

```python

logistic_model.predict_proba(test_X)
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
print(accuracy_score(train_y, y_pred_train))
print(accuracy_score(test_y, y_pred_test))
```

```python
print(confusion_matrix(train_y, y_pred_train))
print('---------------------------------------')
print(confusion_matrix(test_y, y_pred_test))
```

```python
print(classification_report(train_y, y_pred_train))
print('---------------------------------------')
print(classification_report(test_y, y_pred_test))
```

# Hyperparameter Tuning and Cross Validation


```python

model = LogisticRegression()
penalty = ['l1', 'l2', 'elasticnet']
c_value= [100, 10 ,1.0, 0.1, 0.01]
solver = ['newton-cg', 'lbfgs', 'liblinear', 'sag', 'saga']
```

```python
params = dict(penalty = penalty, C = c_value, solver = solver)
```

```python
from sklearn.model_selection import StratifiedKFold
cv = StratifiedKFold()
```

```python
## Grid Search CV
from sklearn.model_selection import GridSearchCV
grid = GridSearchCV(estimator=model, param_grid=params, scoring='accuracy', cv=cv, n_jobs=-1)
```

```python
grid
```

```python
grid.fit(train_X, train_y)
```

```python
grid.best_params_
```

```python
grid.best_score_
```

```python
y_pred = grid.predict(test_X)
```

```python
print(accuracy_score(test_y, y_pred))
```

```python
print(confusion_matrix(test_y, y_pred))
```

```python
print(classification_report(test_y, y_pred))
```

# **Hyperparameter with Randomized Search**

```python
from sklearn.model_selection import RandomizedSearchCV
```

```python
model = LogisticRegression()
randomcv = RandomizedSearchCV(estimator = model, param_distributions=params, cv=5, scoring='accuracy')
```

```python
randomcv.fit(train_X, train_y)
```

```python
randomcv.best_params_
```

```python
randomcv.best_score_
```

```python
y_pred = randomcv.predict(test_X)
```

```python
print(accuracy_score(test_y, y_pred))
print('--------------------------------------')
print(confusion_matrix(test_y, y_pred))
print('--------------------------------------')
print(classification_report(test_y, y_pred))
```
